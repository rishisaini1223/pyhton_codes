abstraction is feature of oops which is used to hide unwanted details from the user.
for achiving abstraction we have to create the abstraction method.

abstract method is method which is not implimented.
to achive abstraction we have to made a abstract class and in this class there should be one abstract method.

abstract class should have 1 abstract method.
here interest_on_amount is abstract method.
abstract method is method which is not implement it will be implememt in child classes.

```python
from abc import ABC, abstractmethod
class RBI:               # it is a abstract class , we cann't made its object.
  document = 'aadhar'   #class variable

  @abstractmethod
  def interest_on_amount(self):
    print('interst will be 8%')

class HDFC(RBI):
  salarySlip= True      #class varible hdfc
  def interest_on_amount(self):
    print('interst will be 15%')
#r1 = RBI()      # Will give error here while creating class

h1 = HDFC()
h1.interest_on_amount()
```

```python
class pythonA25:
  trainer = 'tushar sir'

  def __init__(self,st_name,st_marks):
    self.st_name = st_name
    self.st_marks = st_marks

  def std_info(self):   #instance method
    print(self.trainer, self.st_name)

  @staticmethod         #static method
  def batch_info():
    print('start in january')

  @classmethod          #class method
  def trainer_info(cls):
    print('trainer ', cls.trainer)

class pythonA23(pythonA25):
  trainer =  'vinay sir'
  def __init__(self,pst_name, pst_marks):
    super().__init__(pst_name,pst_marks)

ps1 = pythonA23('naina',67)
print(ps1.trainer_info())


s1 = pythonA25('yash',45)
s1.trainer = 'yz sir'
s1.std_info()
s1.batch_info()
pythonA23.batch_info()
```
